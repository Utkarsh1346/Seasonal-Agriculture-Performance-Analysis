```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('seasonal_agriculture_performance_dataset.csv')
print(df.shape)
print('Duplicates:', df.duplicated().sum())
print('Missing values:\n', df.isnull().sum()[df.isnull().sum()>0])
```

```python
df_clean=df.copy()
for c in ['Rainfall_mm','Soil_Moisture_pct','Yield_Tonnes_Ha']:
    df_clean[c]=df_clean[c].fillna(df_clean[c].median())
print('Remaining missing values:',df_clean.isnull().sum().sum())
```

```python
seasonal_summary=df_clean.groupby('Season').agg(Average_Yield=('Yield_Tonnes_Ha','mean'),Total_Production=('Production_Tonnes','sum'),Average_Profit=('Profit_INR','mean'),Water_Efficiency=('Water_Efficiency_t_per_1000m3','mean'),Disease_Risk=('Disease_Pest_Risk_pct','mean')).reindex(['Kharif','Rabi','Zaid']).round(2)
display(seasonal_summary)
```

```python
irrigation_summary=df_clean.groupby('Irrigation_Method').agg(Average_Profit=('Profit_INR','mean'),Water_Efficiency=('Water_Efficiency_t_per_1000m3','mean')).sort_values('Average_Profit',ascending=False)
display(irrigation_summary.round(2))
```

```python
crop_summary=df_clean.groupby('Crop').agg(Total_Production=('Production_Tonnes','sum'),Average_Profit=('Profit_INR','mean')).sort_values('Average_Profit',ascending=False)
display(crop_summary.round(2))
```

```python
state_summary=df_clean.groupby('State').agg(Total_Production=('Production_Tonnes','sum'),Average_Profit=('Profit_INR','mean')).sort_values('Total_Production',ascending=False)
display(state_summary.round(2))
```
